```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
log_path = "/mnt/d/projects/CysecLLM/ContinuedPretrainingLLM/logs/logs_Qwen2.5-3B-Instruct-CYSECFINETUNED-lora-256/resource_usage_Qwen2.5-3B-Instruct-CYSECFINETUNED-lora-256.csv"
log_df = pd.read_csv(log_path)
```

```python
x = log_df['elapsed_s']
y_ram = log_df['ram_used_gb']
y_cpu = log_df['cpu_percent']
y_gpu = log_df['gpu_mem_used_gb']

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Resource Usage - A100 GPU - Qwen2.5-3B-Instruct', fontsize=16)

# RAM Usage sub-plot
ax1.plot(x, y_ram)
ax1.set_title('RAM Usage Over Time')
ax1.set_xlabel('Elapsed Time (s)')
ax1.set_ylabel('RAM Used (GB)')

# CPU Usage sub-plot
ax2.plot(x, y_cpu, color='orange')
ax2.set_title('CPU Usage Over Time')
ax2.set_xlabel('Elapsed Time (s)')
ax2.set_ylabel('CPU Usage (%)')

# GPU Memory Usage sub-plot
ax3.plot(x, y_gpu, color='green')
ax3.set_title('GPU Memory Usage Over Time')
ax3.set_xlabel('Elapsed Time (s)')
ax3.set_ylabel('GPU Memory Used (GB)')

plt.tight_layout()
plt.show()
```
